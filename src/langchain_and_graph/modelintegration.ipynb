```python
import langchain
```

```python
import os
from dotenv import load_dotenv
load_dotenv()


os.environ["GEMINI_API_KEY"]=os.getenv("GEMINI_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
model = init_chat_model(
    "gemini-3.5-flash",
    model_provider="google_genai",
)
```

```python
model
```

```python
response=model.invoke("Whats the meaning of wassup")
```

```python
response.content
```

```python
for chunk in model.stream("Whats the use of flush in print in py language"):
    print(chunk.content,end="  ",flush=True)
```

```python
try:
    response = model.batch([
        "Tell me about LangChain",
        "What is the point of using LangChain?"
    ])

    for item in response:
        print(item.content)
        print("-" * 80)

except Exception as error:
    if "429" in str(error) or "RESOURCE_EXHAUSTED" in str(error):
        print("Gemini quota exceeded. Wait for the quota reset or check billing/API limits.")
    else:
        raise
for i in response:
    print(i)
```
